```python
# SIMULACIÓN DE IA: EVALUADOR DE CREDIBILIDAD PARA PERIODISTAS
# Este programa evalúa el valor de un testimonio de forma interactiva.

print("--- AI PERIODÍSTICA: MÓDULO DE VERIFICACIÓN ---")
print("IA: Saludos, periodista. Analicemos a tu testigo.")

# El sistema utiliza input para capturar datos (instrucción 'leer' en los diagramas) [6]
tipo_testigo = input("Periodista: ¿Es un testigo 'directo', 'indirecto' o un 'perito'?: ")

# Estructura selectiva múltiple (if-elif-else) para ramificar la lógica [7, 8]
if tipo_testigo == "directo":
    # Uso de condicionales anidados para simular profundidad de razonamiento [9, 10]
    metros = float(input("IA: ¿A qué distancia estaba del suceso? (metros): "))

    # Operador relacional para evaluar proximidad física [5, 11]
    if metros <= 5.0:
        print("IA: Testigo ocular clave. Su cercanía física le otorga alta prioridad.")
    elif metros > 5.0 and metros <= 20.0:
        print("IA: Posición aceptable. Requiere descripción detallada del entorno.")
    else:
        print("IA: La distancia es alta (>20m). La visión podría ser imprecisa; busca fotos.")

elif tipo_testigo == "indirecto":
    pruebas = input("IA: ¿Tiene material físico como 'fotos' o 'audios'? (si/no): ")

    # Lógica basada en la existencia de evidencia material
    if pruebas == "si":
        print("IA: El testimonio es de referencia, pero el material digital lo valida.")
    else:
        # Operador de igualdad para el flujo 'sino' [11]
        print("IA: Riesgo de rumorología. No publicar hasta obtener pruebas tangibles.")

elif tipo_testigo == "perito":
    # El sistema evalúa experiencia técnica mediante operadores aritméticos implícitos
    experiencia = int(input("IA: ¿Cuántos años de trayectoria profesional tiene?: "))

    if experiencia >= 10:
        print("IA: Fuente altamente confiable. Su análisis técnico es irrefutable.")
    else:
        print("IA: Experiencia media. Contrasta sus declaraciones con un colegio profesional.")

else:
    # Caso por defecto si la entrada no coincide con las categorías [7, 12]
    print("IA: Error en categorización. Indica si es directo, indirecto o perito.")

print("--- EVALUACIÓN FINALIZADA ---")
# Descripción de los componentes aplicados:
print("Programa realizado por Iram Jimenez 2H NL = 28")
```
